Overall, the correlation analysis indicates that the workload dimensions are strongly interrelated, while performance exhibits only a modest inverse relationship with workload.

## Outlier Detection using the Interquartile Range (IQR)

```python
import numpy as np

# Time variable
data = df["Time"]

min_val = np.min(data)
Q1 = np.percentile(data, 25)
median = np.median(data)
Q3 = np.percentile(data, 75)
max_val = np.max(data)

five_num_summary = [min_val, Q1, median, Q3, max_val]

print("Five Number Summary:")
print(five_num_summary)
```

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Outliers:")
print(outliers)
```

The IQR method identified several potential outliers in the Time variable. These observations were not removed because they represent genuine differences in participants' task completion times rather than data entry errors. Therefore, all observations were retained for subsequent statistical analyses.

```python
# Select the Time variable
y = df["Time"]
```

```python
# Plot Histogram with KDE to check normality
plt.figure(figsize=(10, 6))
sns.histplot(y, kde=True, bins=30, stat="density", linewidth=0)

plt.title("Histogram with KDE for Time")
plt.xlabel("Time")
plt.ylabel("Density")
plt.show()
```

The normality of the Time variable was examined using a histogram with a Kernel Density Estimate (KDE). This helps to assess whether the distribution is positively skewedв Therefore, the Time variable does not appear to follow a normal distribution.

```python
# Q-Q Plot to check normality visually
import scipy.stats as stats

plt.figure(figsize=(10, 6))
stats.probplot(y, dist="norm", plot=plt)
plt.title('Q-Q Plot for Normality')
plt.show()
```

The Q-Q plot shows noticeable deviations from the reference line, particularly in the upper tail of the distribution. This indicates that the Time variable is not normally distributed. The results are consistent with the histogram and KDE plot, which also suggest a positively skewed distribution.

## One-Way ANOVA

```python
from scipy import stats

group1 = df[df["DB"] == 1]["performance"]
group2 = df[df["DB"] == 2]["performance"]
group3 = df[df["DB"] == 3]["performance"]
group4 = df[df["DB"] == 4]["performance"]
group5 = df[df["DB"] == 5]["performance"]
group6 = df[df["DB"] == 6]["performance"]
group7 = df[df["DB"] == 7]["performance"]
group8 = df[df["DB"] == 8]["performance"]

f_stat, p_value = stats.f_oneway(
    group1,
    group2,
    group3,
    group4,
    group5,
    group6,
    group7,
    group8
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
from scipy import stats

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(group1, group2, group3, group4, group5, group6, group7, group8)

# Display the F-statistic and p-value
f_stat, p_value
```

```python
# Set significance level (alpha)
alpha = 0.05

# Compare p-value with significance level to make a decision
if p_value <= alpha:
    result = f"Reject the null hypothesis: There is a significant difference between the means of the groups.\nF-statistic: {f_stat}\np-value: {p_value}"
else:
    result = f"Fail to reject the null hypothesis: There is no significant difference between the means of the groups.\nF-statistic: {f_stat}\np-value: {p_value}"

# Display the decision
result
```

The One-Way ANOVA test was conducted to compare the mean performance scores across the eight dashboard designs.

The ANOVA results showed:

- **F-statistic = 6.34**
- **p-value = 2.61**

Since the p-value is less than the significance level of 0.05, the null hypothesis (H₀) is rejected.

This indicates that there is a statistically significant difference in the mean performance scores among the dashboard designs. Therefore, dashboard design has a significant effect on user performance.
